## 🧠 Course 1: Foundations of Deep Learning and Neural Networks

### Neural Network Essentials

**Forward propagation** (layer $l$):
$$\mathbf{z}^{[l]} = \mathbf{W}^{[l]} \mathbf{a}^{[l-1]} + \mathbf{b}^{[l]}, \qquad \mathbf{a}^{[l]} = g^{[l]}(\mathbf{z}^{[l]})$$

### Activation Functions

| Function | Formula | Gradient | Use Case |
|----------|---------|----------|---------|
| **ReLU** | $\max(0, z)$ | $\mathbb{1}[z>0]$ | Hidden layers (default) ✅ |
| **Sigmoid** | $\frac{1}{1+e^{-z}}$ | $\sigma(1-\sigma)$ | Binary output |
| **Softmax** | $\frac{e^{z_j}}{\sum_k e^{z_k}}$ | Complex | Multiclass output |
| **Tanh** | $\frac{e^z-e^{-z}}{e^z+e^{-z}}$ | $1-\tanh^2$ | Hidden layers (legacy) |

### Gradient Descent (Adam — default optimizer)

$$m_t = \beta_1 m_{t-1} + (1-\beta_1)\nabla\mathcal{L} \quad \text{(1st moment)}$$
$$v_t = \beta_2 v_{t-1} + (1-\beta_2)(\nabla\mathcal{L})^2 \quad \text{(2nd moment)}$$
$$\theta \leftarrow \theta - \frac{\alpha \hat{m}_t}{\sqrt{\hat{v}_t} + \epsilon}$$

### CNN Key Formulas

$$W_{\text{out}} = \left\lfloor\frac{W + 2p - k}{s}\right\rfloor + 1$$

### Training Toolkit

| Technique | Purpose | Key Formula |
|-----------|---------|-------------|
| **He Init** | Proper scale for ReLU | $W \sim \mathcal{N}(0, \sqrt{2/n_{in}})$ |
| **Batch Norm** | Stabilize distributions | $\hat{x} = \frac{x-\mu_B}{\sqrt{\sigma_B^2+\epsilon}}$, then scale+shift |
| **Dropout** | Prevent co-adaptation | Zero neurons with prob $p$; scale by $\frac{1}{1-p}$ at train |
| **L2 Reg** | Penalize large weights | $\mathcal{L} + \frac{\lambda}{2m}\|\mathbf{W}\|^2$ |

### Seminal CNN Architectures Timeline

```
LeNet-5 (1998) → AlexNet (2012) → VGG (2014) → GoogLeNet (2014) → ResNet (2015)
                  ↑                  ↑                ↑                  ↑
               ReLU/Dropout     Depth+3×3conv    Inception mod.    Skip connections
```

### Key Takeaways ✅
- ReLU + He init + BatchNorm = the modern training default
- Skip connections (ResNet) solve degradation for arbitrarily deep networks
- Backprop = chain rule through a computational graph

## 🔭 Course 2: Advanced Architectures for Computer Vision

### Transfer Learning Decision Framework

```
Dataset size?  Small (<5K) ─── Domain similar? ─── Yes → Feature extraction
               Medium (10K) ─────────────────────────── Progressive unfreeze
               Large (>100K) ───────────────────────────── Full fine-tuning
```

### Transfer Learning Key Concepts

| Concept | Description |
|---------|-------------|
| **Feature Extraction** | Freeze backbone, train new head only |
| **Fine-Tuning** | Unfreeze layers with low LR: $\alpha_{\text{backbone}} \ll \alpha_{\text{head}}$ |
| **Catastrophic Forgetting** | Overwriting pre-trained features with large updates |
| **Domain Shift** | Mismatch between pre-training and target domain distributions |

**Rule of thumb:** Pre-trained weights capture *universal* features (edges, textures) in early layers — these are almost always worth preserving.

### Autoencoder Framework

$$\mathbf{x} \xrightarrow{f_\phi} \mathbf{z} \xrightarrow{g_\theta} \hat{\mathbf{x}}, \qquad \mathcal{L} = \|\mathbf{x} - \hat{\mathbf{x}}\|^2$$

| Variant | Extra Term | Benefit |
|---------|-----------|---------|
| **Undercomplete** | Small bottleneck | Compact features |
| **Denoising** | Corrupt input, clean target | Robust to noise |
| **Sparse** | $\lambda\|\mathbf{z}\|_1$ | Interpretable activations |
| **Contractive** | Jacobian penalty | Locally invariant representations |
| **VAE** | KL divergence to $\mathcal{N}(0,I)$ | Generative, smooth latent space |

**Anomaly detection**: train on normal data only. At test time, anomalies have **high reconstruction error** because the model never learned their structure.

### Key Takeaways ✅
- Transfer learning dramatically reduces data requirements
- Differential learning rates + progressive unfreezing = best fine-tuning practice
- Autoencoders learn meaningful representations without labels
- Reconstruction error is a direct anomaly score

## ⚡ Course 3: Generative Models and Sequence Processing

### GAN Essentials

$$\min_G \max_D \; \mathbb{E}[\log D(\mathbf{x})] + \mathbb{E}[\log(1-D(G(\mathbf{z})))]$$

| Component | Role | Goal |
|-----------|------|------|
| **Generator G** | Creates fake samples from noise $\mathbf{z}$ | Fool D: $D(G(\mathbf{z})) \to 1$ |
| **Discriminator D** | Detects real vs fake | $D(\mathbf{x}_{\text{real}}) \to 1$, $D(G(\mathbf{z})) \to 0$ |

**Failure modes:** Mode collapse, non-convergence → fix with WGAN, progressive growing, label smoothing.

### Sequence Model Comparison

| Model | Memory | Parallelizable | Max Path Length | Notes |
|-------|--------|---------------|----------------|-------|
| **RNN** | $\mathbf{h}_t$ | ❌ | $O(T)$ | Vanishing gradient |
| **LSTM** | $\mathbf{h}_t, \mathbf{c}_t$ (gates) | ❌ | $O(T)$ | Long-range dependencies |
| **GRU** | $\mathbf{h}_t$ (2 gates) | ❌ | $O(T)$ | Faster, simpler LSTM |
| **Transformer** | Attention over all | ✅ | $O(1)$ | Standard for NLP/vision |

### The Attention Formula

$$\text{Attention}(\mathbf{Q},\mathbf{K},\mathbf{V}) = \text{softmax}\!\left(\frac{\mathbf{QK}^T}{\sqrt{d_k}}\right)\mathbf{V}$$

### Transformer Building Blocks

| Block | Function | Formula |
|-------|----------|---------|
| **Multi-Head Attention** | Parallel attention from different subspaces | $\text{Concat}(\text{head}_1,\ldots,\text{head}_h)\mathbf{W}^O$ |
| **Positional Encoding** | Inject position info (no recurrence) | $PE_{pos,2i} = \sin(pos/10000^{2i/d})$ |
| **FFN** | Position-wise non-linearity | $\max(0, \mathbf{x}W_1+b_1)W_2+b_2$ |
| **Residual + LayerNorm** | Stable gradient flow | $\text{LN}(\mathbf{x}+\text{Sublayer}(\mathbf{x}))$ |

### Modern Models at a Glance

| Model | Architecture | Pre-training | Strength |
|-------|-------------|-------------|---------|
| **BERT** | Encoder | MLM + NSP | Understanding, classification |
| **GPT** | Decoder (causal) | Autoregressive LM | Text generation |
| **ViT** | Encoder | Supervised (patches) | Vision at scale |
| **T5** | Encoder-Decoder | Text-to-text | Many NLP tasks unified |

### Key Takeaways ✅
- GANs use adversarial training; training stability is the main challenge
- LSTMs/GRUs gate information — enabling memory over 100s of steps
- Attention directly connects any two positions in $O(1)$ steps
- Transformers = attention + positional enc + FFN + residuals + layer norm
- BERT understands; GPT generates

## 📋 Master Cheat Sheet: Formulas at a Glance

### Loss Functions

| Task | Loss | Formula |
|------|------|---------|
| Regression | MSE | $\frac{1}{m}\sum(\hat{y}-y)^2$ |
| Binary classification | BCE | $-[y\log\hat{y}+(1-y)\log(1-\hat{y})]$ |
| Multiclass | Cross-entropy | $-\sum_k y_k \log \hat{y}_k$ |
| Reconstruction | MSE / BCE | Same as above |
| GAN (discriminator) | BCE | $-[\log D(\mathbf{x}) + \log(1-D(G(\mathbf{z})))]$ |

### Complexity Comparison

| Operation | Time Complexity | Space Complexity |
|-----------|----------------|-----------------|
| Self-Attention | $O(T^2 \cdot d)$ | $O(T^2)$ |
| Recurrent (RNN/LSTM) | $O(T \cdot d^2)$ | $O(d)$ |
| 1D Convolution | $O(T \cdot k \cdot d)$ | $O(k \cdot d)$ |

### Architecture Design Cheat Sheet

| Goal | Component |
|------|-----------|
| Extract visual features | Convolutional layers |
| Reduce spatial dimensions | Pooling / strided convolution |
| Process sequences | LSTM / Transformer |
| Generate data | GAN / VAE / Diffusion |
| Compress to latent space | Autoencoder encoder |
| Transfer knowledge | Pre-trained backbone + new head |
| Stable deep training | BatchNorm + residual connections |
| Prevent overfitting | Dropout, L2, data augmentation |

```
torch.Tensor operations → nn.Module → Custom Dataset/DataLoader
→ Training loops → torchvision → Lightning (for clean boilerplate)
```
