## 📚 Course 1: Supervised Machine Learning

### Key Algorithms

**1. Linear Regression**
- **Model**: $f_{w,b}(x) = wx + b$ (single variable) or $f_{w,b}(\mathbf{x}) = \mathbf{w}^T\mathbf{x} + b$ (multiple variables)
- **Cost Function**: Mean Squared Error (MSE)
  $$J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} (f_{w,b}(x^{(i)}) - y^{(i)})^2$$
- **Use Case**: Predicting continuous values (house prices, stock prices, temperature)

**2. Logistic Regression**
- **Model**: $f_{w,b}(\mathbf{x}) = \frac{1}{1 + e^{-(\mathbf{w}^T\mathbf{x} + b)}}$ (sigmoid function)
- **Cost Function**: Binary Cross-Entropy (Log Loss)
  $$J(w,b) = -\frac{1}{m}\sum_{i=1}^{m}[y^{(i)}\log(f) + (1-y^{(i)})\log(1-f)]$$
- **Use Case**: Binary classification (spam/not spam, disease/healthy)

### Core Concepts

| Concept | Description | Key Formula |
|---------|-------------|-------------|
| **Gradient Descent** | Optimization algorithm to minimize cost | $w := w - \alpha \frac{\partial J}{\partial w}$ |
| **Learning Rate** ($\alpha$) | Step size in gradient descent | Typical: 0.001 - 0.3 |
| **Feature Scaling** | Normalize features to similar ranges | Z-score: $x' = \frac{x - \mu}{\sigma}$ |
| **Regularization** | Prevent overfitting by penalizing large weights | $J_{reg} = J + \frac{\lambda}{2m}\sum w_j^2$ |
| **Polynomial Features** | Create non-linear models | Add $x^2, x^3, \sqrt{x}$, etc. |

### Key Takeaways
✅ Linear regression for continuous predictions  
✅ Logistic regression for binary classification  
✅ Gradient descent is a general optimization technique  
✅ Feature scaling speeds up convergence  
✅ Regularization prevents overfitting

## 🧠 Course 2: Advanced Learning Algorithms

### Neural Networks

**Architecture Components:**
- **Input Layer**: Receives features
- **Hidden Layers**: Learn intermediate representations
- **Output Layer**: Produces predictions

**Forward Propagation:**  
For each layer $l$:
1. $z^{[l]} = \mathbf{W}^{[l]} a^{[l-1]} + \mathbf{b}^{[l]}$
2. $a^{[l]} = g^{[l]}(z^{[l]})$ (apply activation function)

### Activation Functions

| Function | Formula | Use Case | Range |
|----------|---------|----------|-------|
| **ReLU** | $\max(0, z)$ | Hidden layers (most common) | $[0, \infty)$ |
| **Sigmoid** | $\frac{1}{1+e^{-z}}$ | Binary classification output | $(0, 1)$ |
| **Softmax** | $\frac{e^{z_j}}{\sum e^{z_k}}$ | Multiclass output | $(0, 1)$, sum=1 |
| **Tanh** | $\frac{e^z - e^{-z}}{e^z + e^{-z}}$ | Hidden layers (less common) | $(-1, 1)$ |
| **Linear** | $z$ | Regression output | $(-\infty, \infty)$ |

**TensorFlow Basics:** Define → Compile → Train → Predict

**Multiclass:** Use Softmax activation with `SparseCategoricalCrossentropy(from_logits=True)`

### Practical Advice

**Bias vs Variance:**

| Problem | Symptoms | Solutions |
|---------|----------|-----------|
| **High Bias** (Underfitting) | $J_{train}$ high, $J_{cv}$ high | More features, decrease $\lambda$, bigger network |
| **High Variance** (Overfitting) | $J_{train}$ low, $J_{cv}$ high | More data, increase $\lambda$, regularization |

**Data Splits:**
- Training set (60-80%): Train model
- Validation/Cross-validation set (10-20%): Tune hyperparameters
- Test set (10-20%): Final evaluation (use once!)

### Decision Trees & Ensembles

**Decision Trees:**
- Split data based on features
- Use entropy/information gain to choose best splits
- Prone to overfitting

**Random Forest:**
- Ensemble of many decision trees
- Each tree trained on random sample of data
- Reduces overfitting, more robust

**XGBoost:**
- Gradient boosting (sequential trees correct previous errors)
- State-of-the-art for tabular data
- Hyperparameters: `n_estimators`, `learning_rate`, `max_depth`

**When to Use What:**
- **Tabular/Structured Data**: XGBoost or Random Forest
- **Images/Audio/Text**: Neural Networks

### Key Takeaways
✅ Neural networks learn hierarchical features  
✅ ReLU for hidden layers, sigmoid/softmax for output  
✅ TensorFlow makes building NNs easy  
✅ Diagnose bias/variance before deciding what to try  
**Tree Models:** Decision Trees → Random Forest (ensemble) → XGBoost (gradient boosting)
- **Use for**: Tabular/structured data
- **Neural Networks for**: Images, audio, text

## 🔍 Course 3: Unsupervised Learning, Recommenders, RL

### 1. Clustering (K-Means)

**Algorithm:**
1. **Initialize**: Pick $K$ random centroids
2. **Assignment**: Assign each point to nearest centroid
3. **Update**: Move centroids to mean of assigned points
4. **Repeat** until convergence

**Cost Function (Distortion):**
$$J = \frac{1}{m}\sum_{i=1}^{m} ||x^{(i)} - \mu_{c^{(i)}}||^2$$

**Tips:**
- Run multiple times with different initializations
- Use Elbow method to choose $K$ (but not always clear)
- Choose $K$ based on application needs

### 2. Anomaly Detection

**Gaussian Distribution Model:**
$$p(x) = \prod_{j=1}^{n} \frac{1}{\sqrt{2\pi\sigma_j^2}} e^{-\frac{(x_j-\mu_j)^2}{2\sigma_j^2}}$$

**Algorithm:**
1. Fit Gaussian parameters $\mu_j, \sigma_j^2$ for each feature
2. Compute $p(x)$ for new example
3. Flag as anomaly if $p(x) < \epsilon$

**When to Use:**
- Very few positive examples (anomalies)
- Many negative examples (normal)
- Future anomalies may look different from current ones
- Examples: Fraud detection, manufacturing defects

### 3. Recommender Systems

**Collaborative Filtering:** Learn from user-item interactions ("Users like you also liked...")

**Content-Based:** Use features of users/items with neural networks (can recommend new items immediately)

### 4. Reinforcement Learning

**Key Concepts:**

| Term | Definition |
|------|------------|
| **State** ($s$) | Current situation |
| **Action** ($a$) | What agent can do |
| **Reward** ($R$) | Feedback signal |
| **Policy** ($\pi$) | Strategy for choosing actions |
| **Return** | Total discounted reward: $R_1 + \gamma R_2 + \gamma^2 R_3 + ...$ |
| **Q-function** | $Q(s,a)$ = Expected return from state $s$, taking action $a$ |

**Bellman Equation:**
$$Q(s,a) = R(s,a) + \gamma \max_{a'} Q(s',a')$$

**Deep Q-Networks (DQN):**
- Use neural network to approximate $Q(s,a)$
- Experience replay: Store and sample past experiences
- Target network: Stabilize training
- $\epsilon$-greedy: Balance exploration vs exploitation

**When to Use RL:**
- Sequential decision making
- No direct supervision
- Long-term consequences
- Need to explore to find good strategies

### Key Takeaways
✅ K-Means finds natural groupings in data  
✅ Anomaly detection identifies rare events  
✅ Collaborative filtering learns from user behavior  
✅ Content-based uses features for recommendations  
✅ RL learns optimal behavior through trial and error

## 🎯 Algorithm Selection Guide

### Choose Your Algorithm Based on the Problem Type

```
┌─────────────────────────────────────────────┐
│         Do you have labeled data?           │
└──────────────┬──────────────────────────────┘
               │
       ┌───────┴───────┐
       │               │
      YES             NO
       │               │
       ▼               ▼
  SUPERVISED    UNSUPERVISED
       │               │
       │               ├─ Grouping → K-Means Clustering
       │               ├─ Anomalies → Anomaly Detection
       │               └─ Recommendations → Collaborative/Content-Based
       │
       ├─ Continuous Output → Regression
       │   ├─ Linear relationship → Linear Regression
       │   ├─ Non-linear → Polynomial Regression
       │   └─ Complex patterns → Neural Network
       │
       └─ Discrete Output → Classification
           ├─ Binary (2 classes) → Logistic Regression
           ├─ Multiclass → Softmax Regression
           ├─ Complex patterns → Neural Network
           └─ Tabular data → Random Forest / XGBoost
```

### Data Type Guide

| Data Type | Recommended Approaches |
|-----------|------------------------|
| **Tabular/Structured** | Linear/Logistic Regression, Random Forest, XGBoost |
| **Images** | Convolutional Neural Networks (CNNs) |
| **Text** | Neural Networks, Transformers |
| **Audio** | Neural Networks, RNNs |
| **Time Series** | LSTM, RNNs, ARIMA |
| **Sequential Decisions** | Reinforcement Learning |

## 📊 Key Formulas Reference

### Loss Functions

| Task | Loss Function | Formula |
|------|---------------|---------|
| **Regression** | Mean Squared Error | $\frac{1}{2m}\sum(f(x^{(i)}) - y^{(i)})^2$ |
| **Binary Classification** | Binary Cross-Entropy | $-\frac{1}{m}\sum[y\log(f) + (1-y)\log(1-f)]$ |
| **Multiclass** | Categorical Cross-Entropy | $-\frac{1}{m}\sum\sum y_j \log(a_j)$ |

### Regularization

**L2 Regularization (Ridge):**
$$J_{regularized} = J_{original} + \frac{\lambda}{2m}\sum_{j=1}^{n}w_j^2$$

**Effect on Gradient Descent:**
$$w_j := w_j(1 - \alpha\frac{\lambda}{m}) - \alpha\frac{\partial J}{\partial w_j}$$

### Evaluation Metrics

| Metric | Formula | Use Case |
|--------|---------|----------|
| **Accuracy** | $\frac{\text{Correct}}{\text{Total}}$ | Balanced classes |
| **Precision** | $\frac{TP}{TP + FP}$ | Cost of false positives |
| **Recall** | $\frac{TP}{TP + FN}$ | Cost of false negatives |
| **F1-Score** | $\frac{2 \cdot P \cdot R}{P + R}$ | Imbalanced classes |
